```python
import numpy as np

# Marks of 4 students in 4 subjects
# Order: Math, Science, English, History
student_scores = np.array([
    [85, 90, 78, 88],
    [92, 86, 80, 75],
    [78, 95, 85, 82],
    [88, 89, 92, 90]
])

# Calculate average score for each subject
average_scores = np.mean(student_scores, axis=0)

# Display average scores
subjects = ["Math", "Science", "English", "History"]

for i in range(4):
    print(subjects[i], "Average:", average_scores[i])

# Find the subject with the highest average
highest_index = np.argmax(average_scores)

print("Subject with highest average:", subjects[highest_index])
print("Highest average score:", average_scores[highest_index])
```
