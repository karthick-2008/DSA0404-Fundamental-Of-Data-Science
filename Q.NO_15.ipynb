```python
import pandas as pd

# Create a dataset
data = {'Likes': [10, 20, 10, 30, 20, 10, 40, 30, 20, 10]}

df = pd.DataFrame(data)

# Calculate frequency distribution
frequency = df['Likes'].value_counts().sort_index()

# Display the result
print("Frequency Distribution of Likes:")
print(frequency)
```
