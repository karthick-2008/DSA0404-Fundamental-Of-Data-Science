```python

import pandas as pd
from google.colab import files

# Step 1: Upload dataset
uploaded = files.upload()

# Step 2: Read Excel file
filename = list(uploaded.keys())[0]
df = pd.read_excel(filename)

# Step 3: Display dataset
print("Dataset:")
print(df.head())

# Step 4: Calculate purchase cost
subtotal = (df['Price'] * df['Quantity_Sold']).sum()

# Step 5: Calculate discount
discount = (subtotal * df['Discount']).sum() / len(df)

# Step 6: Calculate amount after discount
amount = subtotal - discount

# Step 7: Calculate 5% tax
tax_rate = 5
tax = amount * tax_rate / 100

# Step 8: Calculate final cost
total_cost = amount + tax

# Step 9: Display result
print("\n----- CUSTOMER BILL -----")
print("Subtotal:", round(subtotal, 2))
print("Discount:", round(discount, 2))
print("Tax (5%):", round(tax, 2))
print("Total Cost:", round(total_cost, 2))
```
