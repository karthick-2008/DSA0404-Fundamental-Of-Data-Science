```python
from google.colab import files
uploaded = files.upload()
import pandas as pd

# Read dataset
df = pd.read_excel("Sales (1) (1).xlsx")

# Select first 5 products as a sample purchase
prices = df["Price"].head(5).tolist()
quantities = df["Quantity_Sold"].head(5).tolist()
discounts = df["Discount"].head(5).tolist()

# Calculate total amount
total = 0

for i in range(len(prices)):
    total += prices[i] * quantities[i]

# Calculate discount
discount = 0

for i in range(len(prices)):
    discount += prices[i] * quantities[i] * discounts[i]

# Amount after discount
amount = total - discount

# Calculate tax
tax_rate = 5
tax = amount * tax_rate / 100

# Calculate final cost
final_cost = amount + tax

# Display results
print("Total Amount:", round(total, 2))
print("Discount:", round(discount, 2))
print("Amount After Discount:", round(amount, 2))
print("Tax:", round(tax, 2))
print("Final Cost:", round(final_cost, 2))
```
