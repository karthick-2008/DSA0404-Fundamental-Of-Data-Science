```python

import pandas as pd
from google.colab import files

# Step 1: Upload dataset
uploaded = files.upload()

# Step 2: Read Excel file
filename = list(uploaded.keys())[0]
order_data = pd.read_excel(filename)

# Step 3: Display dataset
print("Customer Order Dataset:")
print(order_data.head())

# Step 4: Total number of orders by each customer
print("\n1. Total Orders Made by Each Customer:")
total_orders = order_data.groupby('customer_name')['order_id'].nunique()
print(total_orders)

# Step 5: Average order quantity for each product
print("\n2. Average Order Quantity for Each Product:")
average_quantity = order_data.groupby('product_name')['quantity'].mean()
print(average_quantity)

# Step 6: Convert order date into datetime format
order_data['order_date'] = pd.to_datetime(order_data['order_date'])

# Step 7: Find earliest and latest order dates
print("\n3. Earliest Order Date:")
print(order_data['order_date'].min())

print("\nLatest Order Date:")
print(order_data['order_date'].max())
```
